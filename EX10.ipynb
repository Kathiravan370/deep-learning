```python
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Load a dedicated conversational model directly into Colab memory
print("Downloading Chatbot Model... (This takes about 30 seconds)")
model_name = "facebook/blenderbot-400M-distill"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# 2. Define the chat function (handles user input and generates bot response)
def chat_with_bot(user_message, history):
    # Tokenize the user's input
    inputs = tokenizer([user_message], return_tensors="pt")

    # Generate a conversational reply using the model (added do_sample=True to allow temperature)
    reply_ids = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)

    # Decode the response back into readable text
    bot_reply = tokenizer.batch_decode(reply_ids, skip_special_tokens=True)[0]

    return bot_reply

# 3. Build and launch the Chatbot Interface
demo = gr.ChatInterface(
    fn=chat_with_bot,
    title="🤖 Local AI Chatbot (Blenderbot)",
    description="Experiment 10: Building a Simple LLM ChatBot (Running locally on Colab hardware).",
    examples=["Hello! How are you doing today?", "Can you tell me a joke?", "What is your favorite color?"]
)

# Launch with share=True to get a public link you can test in a separate tab
demo.launch(share=True)
```
