```python
# Install first (if needed): pip install pydatalog
from pydatalog import pyDatalog

# Clear old rules/facts
pyDatalog.clear()

# Create atoms (predicates)
pyDatalog.create_atoms('Door, Window, Motion, Home, Alarm, X')

# ---------------------------------------------------------
# Rules (Knowledge Base)
# ---------------------------------------------------------
# Rule 1: If Door is open AND Motion is detected → Alarm
Alarm(X) <= (Door(X) & Motion(X))

# Rule 2: If Window is open → Alarm
Alarm(X) <= Window(X)

# Rule 3: If Motion is detected AND NOT Home → Alarm
Alarm(X) <= (Motion(X) & ~Home(X))

# ---------------------------------------------------------
# Facts (Scenarios)
# ---------------------------------------------------------
# Scenario 1: Door open + Motion detected + No one home
+Door('house1')
+Motion('house1')
-Home('house1')   # negative fact: no one is home

# Scenario 2: Window open only
+Window('house2')
+Home('house2')

# Scenario 3: Motion detected but someone is home
+Motion('house3')
+Home('house3')

# Scenario 4: All safe (someone home, no sensors triggered)
+Home('house4')

# ---------------------------------------------------------
# Queries
# ---------------------------------------------------------
print("Scenario 1 (Door+Motion, Home=False):", Alarm('house1'))
print("Scenario 2 (Window open):", Alarm('house2'))
print("Scenario 3 (Motion but Home=True):", Alarm('house3'))
print("Scenario 4 (All safe):", Alarm('house4'))
```
