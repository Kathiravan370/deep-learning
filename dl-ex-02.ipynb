```python
print("24BAD057 24BAD057")
import numpy as np
import matplotlib.pyplot as plt
print("Part-A")

x = np.linspace(-10, 10, 400)

sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plt.plot(x, sigmoid)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.subplot(3, 1, 2)
plt.plot(x, tanh)
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(x, relu)
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
print("part-B")
print("Kathiravan D-24BAD057")
print("Part B - Performance Comparison of Activation Functions")
import tensorflow as tf
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load MNIST Dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize the pixel values
X_train = X_train / 255.0
X_test = X_test / 255.0

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
activations = ['sigmoid', 'tanh', 'relu']

results = {}
```

```python
for act in activations:

    print("\n====================================")
    print("Training using:", act.upper())
    print("====================================")

    model = tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape=(28,28)),
        tf.keras.layers.Dense(128, activation=act),
        tf.keras.layers.Dense(64, activation=act),
        tf.keras.layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        X_train,
        y_train,
        validation_data=(X_test, y_test),
        epochs=10,
        verbose=1
    )

    results[act] = history.history
```

```python
comparison = []

for act in activations:

    history = results[act]

    comparison.append([
        act.upper(),
        history['accuracy'][-1],
        history['val_accuracy'][-1],
        history['loss'][-1],
        history['val_loss'][-1],
        len(history['accuracy'])
    ])

df = pd.DataFrame(
    comparison,
    columns=[
        "Activation",
        "Training Accuracy",
        "Validation Accuracy",
        "Training Loss",
        "Validation Loss",
        "Epochs"
    ]
)

print(df)
```

```python
plt.figure(figsize=(8,5))

for act in activations:
    plt.plot(results[act]['val_accuracy'], label=act)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))

for act in activations:
    plt.plot(results[act]['val_loss'], label=act)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
best = df.loc[df["Validation Accuracy"].idxmax()]

print("Best Activation Function")
print("------------------------")
print("Activation Function :", best["Activation"])
print("Training Accuracy   :", best["Training Accuracy"])
print("Validation Accuracy :", best["Validation Accuracy"])
```

```python
print("Part c")
print("kathiravan-24BAD057")
import tensorflow as tf
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load MNIST Dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize the data
X_train = X_train / 255.0
X_test = X_test / 255.0

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
optimizers = {
    "SGD": tf.keras.optimizers.SGD(),
    "Momentum": tf.keras.optimizers.SGD(momentum=0.9),
    "RMSProp": tf.keras.optimizers.RMSprop(),
    "Adam": tf.keras.optimizers.Adam()
}

results = {}
```

```python
for name, optimizer in optimizers.items():

    print("\n===================================")
    print("Training using:", name)
    print("===================================")

    model = tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape=(28,28)),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(64, activation='relu'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        X_train,
        y_train,
        epochs=10,
        validation_data=(X_test, y_test),
        verbose=1
    )

    results[name] = history.history
```

```python
comparison = []

for name in optimizers:

    history = results[name]

    comparison.append([
        name,
        history['loss'][-1],
        history['val_loss'][-1],
        history['accuracy'][-1],
        history['val_accuracy'][-1],
        len(history['accuracy'])
    ])

df = pd.DataFrame(
    comparison,
    columns=[
        "Optimizer",
        "Training Loss",
        "Validation Loss",
        "Training Accuracy",
        "Validation Accuracy",
        "Epochs"
    ]
)

print(df)
```

```python
plt.figure(figsize=(8,5))

for name in optimizers:
    plt.plot(results[name]['loss'], label=name)

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

for name in optimizers:
    plt.plot(results[name]['val_loss'], label=name)

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

for name in optimizers:
    plt.plot(results[name]['accuracy'], label=name)

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

for name in optimizers:
    plt.plot(results[name]['val_accuracy'], label=name)

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
best = df.loc[df["Validation Accuracy"].idxmax()]

print("Best Optimizer")
print("--------------------------")
print("Optimizer           :", best["Optimizer"])
print("Training Accuracy   :", best["Training Accuracy"])
print("Validation Accuracy :", best["Validation Accuracy"])
print("Training Loss       :", best["Training Loss"])
print("Validation Loss     :", best["Validation Loss"])
```
